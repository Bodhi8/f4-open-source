# F4 public demonstration

All records and fraud schemes in this notebook are **synthetic**. Install `vector1-f4-core` from this checkout before running.

```python
from f4 import generate_portfolio, Payoffs, expected_values
from f4.contracts import validate_portfolio

portfolio = generate_portfolio(seed=42)
assert validate_portfolio(portfolio.tables) == []
{k: len(v) for k, v in portfolio.tables.items()}
```

## Synthetic scheme composition
Truth below is for evaluation only and is stored separately from invoice inputs.

```python
from collections import Counter

Counter(portfolio.truth.values())
```

```python
from f4.features import basic_graph_features

edges = [(r.carrier_id, r.debtor_id) for r in portfolio.tables["Invoice"]]
basic_graph_features(edges)["c0"]
```

```python
values = expected_values(0.1, Payoffs(100, 1000, 0.8, 10, 50))
assert abs(values["verify"] - 60) < 1e-9
values
```

These expected values follow explicit input assumptions. They do not measure client savings. The public graph features are a reference example, not a fraud classifier.
